```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
```

## Task 2

#### Import Datasets

```python
p2a = pd.read_csv('p2a.csv', header=None, names=['variable1', 'variable2'])
p2b = pd.read_csv('p2b.csv', header=None, names=['variable1', 'variable2'])
p2c = pd.read_csv('p2c.csv', header=None, names=['variable1', 'variable2'])

p2a_var1 = p2a['variable1']
p2a_var2 = p2a['variable2']

p2b_var1 = p2b['variable1']
p2b_var2 = p2b['variable2']

p2c_var1 = p2c['variable1']
p2c_var2 = p2c['variable2']
```

### Part a

```python
correlation_coefficient_a, p_value_a = pearsonr(p2a_var1, p2a_var2)
#Let's set the alpha value as 0.05 If p_value is less than 0.05, reject; else, FTR
print(f"The p value is {p_value_a}")
#ergo, we reject the null hypothesis

#This is the r value, which finds the magnitude
print(f"The r-value is {correlation_coefficient_a}")
#This is a positive, moderate correlation
```

### Part b

```python
correlation_coefficient_b, p_value_b = pearsonr(p2b_var1, p2b_var2)
#Let's set the alpha value as 0.05 If p_value is less than 0.05, reject; else, FTR
print(f"The p value is {p_value_b}")
#ergo, we reject the null hypothesis

#This is the r value, which finds the magnitude
print(f"The r-value is {correlation_coefficient_b}")
#This is a positive, strong correlation
```

```python
print(f"Comparison of correlation coeff.: A: {correlation_coefficient_a:.3f} vs B: {correlation_coefficient_b:.3f}")

print(f"Comparison of p-values: A:{p_value_a} vs B:{p_value_b}")
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(p2a_var1, p2a_var2, alpha=0.6, color = 'blue')
plt.xlabel('Variable 1')
plt.ylabel('Variable 2')
plt.title('Scatter plot for p2a')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(p2b_var1, p2b_var2, alpha=0.6, color = 'green')
plt.xlabel('Variable 1')
plt.ylabel('Variable 2')
plt.title('Scatter plot for p2b')
plt.grid(True)
plt.show()
```

### Part c

```python
correlation_coefficient_c, p_value_c = pearsonr(p2c_var1, p2c_var2)
#Let's set the alpha value as 0.05 If p_value is less than 0.05, reject; else, FTR
print(f"The p value is {p_value_c}")
#ergo, we FTR the null hypothesis

#This is the r value, which finds the magnitude
print(f"The r-value is {correlation_coefficient_c}")
#This is a weak positive correlation
```

```python
print(f"Comparison of correlation coeff.: A: {correlation_coefficient_a:.3f} vs C: {correlation_coefficient_c:.3f}")

print(f"Comparison of p-values: A:{p_value_a} vs C:{p_value_c}")
```
