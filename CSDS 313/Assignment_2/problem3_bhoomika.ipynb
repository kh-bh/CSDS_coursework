```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

##### Dataset

```python
cars = pd.read_csv("cars.csv", names=['US', 'Japan'])
```

```python
US = cars['US']
Japanese = cars['Japan'].dropna()
```

##### Part A

```python
print(f"The Number of US cars in the dataset is {len(cars['US'])}")
print(f"The Number of Japanese cars in the dataset is {cars['Japan'].count()}")
```

##### Part D

```python
stat, p_value = stats.ttest_ind(US, Japanese, equal_var = False)
print(f"t-test: statistic={stat:.4f}, p-value={p_value:.4f}")
```
